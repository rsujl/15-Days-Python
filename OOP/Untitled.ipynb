```python
class Point:
    def __init__(self,x,y):
        self.x=x
        self.y=y
    def __str__(self):
        return '<{},{}>'.format(self.x,self.y)
    def euc_distance(self,other):
        return ((self.x-other.x)**2 + (self.y-other.y)**2)**0.5
    def dis_origin(self):
        return self.euc_distance(Point(0,0))
class Line:
    def __init__(self,a,b,c,):
        self.a=a
        self.b=b
        self.c=c
    def __str__(self):
        return '{}x+{}y+{}'.format(self.a,self.b,self.c)
    def calculate(self,point):
        if self.a *point.x +self.b*point.y+self.c==0:
            return "Lies"
        else:
            return "not"
        
```

```python
class Atm:
    def __init__(self):
        print(f"Object created at memory address: {id(self)}")
        self.pin = ''
        self.__balance = 0
        #self.menu()  # Automatically start the menu when object is created
    def get_balance(self):
        return self.__balance
    def set_balance(self,new_value):
        if type(new_value) ==int:
            self.__balance=new_value
        else:
            print("Nikal Bahar")
        

    def menu(self):
        user_input = input(
            "\n--- ATM Menu ---\n"
            "Press 1 to create Pin \n"
            "Press 2 to change pin \n"
            "Press 3 to check balance \n"
            "Press 4 to withdraw \n"
            "Anything else to Exit \n"
            "Selection: "
        )
        if user_input == "1":
            self.createPin()
        elif user_input == "2":
            self.changePin()
        elif user_input == "3":
            self.checkBalance()
        elif user_input == "4":
            self.withdrawBalance()
        else:
            print("Thank you for using the ATM. Goodbye!")
            return  # Clean exit from the menu loop

    def createPin(self):
        user_pin = input("Enter Pin: ")
        self.pin = user_pin
        user_balance = int(input("Enter initial balance: "))
        self.__balance = user_balance
        print("Pin created Successfully!")
        self.menu()

    def changePin(self):
        old_pin = input("Enter Old pin: ")
        if old_pin == self.pin:
            newPin = input("Enter New Pin: ")
            self.pin = newPin
            print("Pin changed successfully!")
        else:
            print("Incorrect Old Pin!")
        self.menu()

    def checkBalance(self):
        user_pin = input("Enter Pin: ")
        if user_pin == self.pin:
            print("Balance is:", self.__balance)
        else:
            print("Invalid Pin!")
        self.menu()

    def withdrawBalance(self):
        user_pin = input("Enter Pin: ")
        if user_pin == self.pin:
            useramt = int(input("Enter amount to withdraw: "))
            if useramt <= self.__balance:
                self.__balance -= useramt
                print("Amount Debited:", useramt)
                print("Remaining Balance:", self.__balance)
            else:
                print("Insufficient Balance!")
        else:
            print("Invalid Pin!")
        self.menu()
```
