```python
import numpy as np

def create_border_array(a, b):
    # Sabse pehle saare elements ko 1 rakh kar (a, b) shape ka array banayein
    arr = np.ones((a, b), dtype=int)
    
    # Slicing use karke andar ke elements (1st row se last-1 row tak, 
    # aur 1st col se last-1 col tak) ko 0 kar dein
    arr[1:-1, 1:-1] = 0
    
    return arr

# Function ko test karne ke liye input lein
rows = int(input("Enter number of rows (a): "))
cols = int(input("Enter number of columns (b): "))

# Function call karein aur print karein
result = create_border_array(rows, cols)
print("\nGenerated Border Array:")
print(result)
```
