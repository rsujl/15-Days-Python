```python
import json
class Person:
    def __init__(self,name,lname,age,gender):
        self.name=name
        self.lname=lname
        self.age=age
        self.gender=gender
person=Person('sujal','rathos',23,'Male')
def show_object(person):
    if isinstance(person,Person):
       return{'name':person.name + '' + person.lname,'age':person.age,'gender':person.gender}
with open('demo.json','w') as f:
    json.dump(person,f,default=show_object)
```

```python
class Person:
    def __init__(self,name,age):
        self.age=age
        self.name=name
    def display(self):
        print(f"Hi my name is {self.name} and age us {self.age}")
p=Person('sujal',18)
```

```python
import pickle
with open('d.pkl','wb') as f:
    pickle.dump(p,f)
```

```python
import pickle
with open('d.pkl','rb') as f:
    p=pickle.load(f)
p.display()
```

```python
def mul(a,b):
    if b==1:
        return a
    else:
        return a+mul(a,b-1)
mul(2,2)
```

```python
def fact1(n):
    fact=1
    for i in range(1,n+1):
        fact=fact*i
    print(fact)
fact1(5)
```

```python
def factorial(n):
    fact==1
    if n==1:
        return n
    else:
        return n*factorial(n-1)
factorial(5)
```

```python
def palindrome(str):
    if len(str)==1:
        print("Is palindrome")
    else:
        if str[0]==str[-1]:
            palindrome(str[1:-1])
        else:
            print("Not palndrome")
palindrome("madam")
```
