```python
def sujal(func):
    def wrapper():
        print("***************")
        func()
        print("***************")
    return wrapper
@sujal
def hello():
    print("Hello")
hello()
```

```python
import time
def timer(func):
    def wrapper(*args):
        start=time.time()
        func(*args)
        print("Time takes is",func.__name__,time.time()-start)
    return wrapper
@timer
def hello(a,b):
    print(a*b)
@timer
def mul(a,b):
    time.sleep(10)
    print(a-b)
hello(3,4)
mul(3,4)
```

```python
def sanity_check(data_type):
    def outer_wrapper(func):
        def inner_wrapper(*args):
            if type(*args) == data_type:
                func(*args)
            else:
                raise TypeError("Will not work")
        return inner_wrapper
    return outer_wrapper

@sanity_check(int)
def square(num):
    print(num**2)
square(3)
```
